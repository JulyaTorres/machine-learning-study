# DBSCAN

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
```

```python
X, y = make_moons(n_samples=300, noise=0.05)
```

```python
# Modelo
dbscan = DBSCAN(eps=0.2, min_samples=5) # eps -> raio de vizinhaça que escolhemos aceitar, se 2 pontos são vizinhos, eles são vizinhos a partir de qual distância euclidiana máxima 
dbscan.fit(X)
```

```python
y_db = dbscan.fit_predict(X)
print(f"Número de clusters: {len(set(y_db))}")
```

```python
#Visualização
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=y_db, cmap='viridis')
plt.title("DBSCAN Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.grid(True)
plt.show()
```
