# Regras de Associação

## Apriori

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
#transações os itens
transacoes = [
    ['leite', 'pão', 'manteiga'],
    ['café', 'pão', 'café'],
    ['leite', 'manteiga','pão'],
    ['café', 'pão', 'manteiga','café'],
    ['leite', 'café', 'pão', 'manteiga']
]
```

```python
# Codificando as transações (one-hot encoding)
te = TransactionEncoder()
dados_transformados = te.fit(transacoes).transform(transacoes)
df = pd.DataFrame(dados_transformados, columns=te.columns_)
```

```python
print("Dados codificados:")
print(df)
```

```python
# Modelo
frequentes = apriori(df, min_support=0.6, use_colnames=True)
```

```python
print("Itens frequentes (com suporte mínimo de 0.6):")
print(frequentes)
```

```python
regras = association_rules(frequentes, metric="confidence", min_threshold=0.7)
```

```python
print("Regras de Associação:")
print(regras[["antecedents","consequents","support","confidence","lift"]])
```
