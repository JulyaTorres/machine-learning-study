# Clusterização

## Kmeans

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs # gerar dados artificiais
from sklearn.cluster import KMeans 
```

```python
X, y_true = make_blobs(n_samples=300, centers=3, cluster_std=0.6, random_state=42) 
```

```python
#Modelo
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X)
y_kmeans = kmeans.predict(X)
```

```python
#Inercia
print(f"Inércia: {kmeans.inertia_}")
```

```python
# Visualização

plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=y_kmeans, cmap='viridis')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], c='red', label='Centros')
plt.title("K-Means com 3 Clusters")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.grid(True)
plt.show()
```

## Método do Cotovelo

```python
inertias = []
range =  range(1, 11)
for k in range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)
    inertias.append(kmeans.inertia_)
```

```python
# Visualização método do cotovelo
plt.figure(figsize=(8, 6))
plt.plot(range, inertias, marker='o')
plt.title("Método do Cotovelo")
plt.xlabel("Número de Clusters (k)")
plt.ylabel("Inércia")
plt.xticks(range)
plt.grid(True)
plt.show()
```
