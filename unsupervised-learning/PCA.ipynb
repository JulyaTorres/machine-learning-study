# Redução de dimencionalidade

## PCA

```python
import matplotlib.pyplot as plt
import mpl_toolkits.mplot3d.axes3d
from sklearn.decomposition import PCA
from sklearn.datasets import load_iris
import pandas as pd
```

```python
iris = load_iris()
X = iris.data
y = iris.target
target_names = iris.target_names
```

```python
print(target_names)
```

```python
# PCA
pca = PCA(n_components=3)
X_reduced = pca.fit(X).transform(X)
```

```python
# Visualização
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection="3d", elev=-150, azim=110)

scatter = ax.scatter(
    X_reduced[:, 0],
    X_reduced[:, 1],
    X_reduced[:, 2],
    c=y,
    cmap='viridis',
    s=40
)

ax.set_title("Três primeiras componentes principais (PCA) – Iris")
ax.set_xlabel("1ª Componente Principal")
ax.set_ylabel("2ª Componente Principal")
ax.set_zlabel("3ª Componente Principal")
ax.xaxis.set_ticklabels([])
ax.yaxis.set_ticklabels([])
ax.zaxis.set_ticklabels([])
legend1 = ax.legend(
    scatter.legend_elements()[0],
    target_names,
    loc="upper right",
    title="Espécies"
)
ax.add_artist(legend1)

# Verificar os componentes e as features
componentes = pd.DataFrame(
    pca.components_,
    columns=iris.feature_names,
    index=["PC1", "PC2", "PC3"]
)
print(componentes)
```
