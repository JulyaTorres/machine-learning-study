```python
import pandas as pd
from pyECLAT import ECLAT
```

```python
transacoes = [
    ['leite', 'pão', 'manteiga'],
    ['café', 'pão', 'café'],
    ['leite', 'manteiga','pão'],
    ['café', 'pão', 'manteiga','café'],
    ['leite', 'café', 'pão', 'manteiga']
]
```

```python
df = pd.DataFrame(transacoes)
```

```python
eclat = ECLAT(data=df)
```

```python
frequentes, indices = eclat.fit(min_support=0.6, min_combination=1, max_combination=3)
```

```python
print("Itens frequentes (com suporte mínimo de 0.6):")
print(frequentes)
```

```python
indices, suporte = eclat.fit(min_support=0.6, min_combination=1, max_combination=3)
```

```python
print("\n Itens com suporte mínimo de 0.6:")
for item, sup in suporte.items():
    print(f"{item}: {sup:.2f}")
```
