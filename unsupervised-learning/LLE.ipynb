## LLE

```python
from sklearn.manifold import LocallyLinearEmbedding
from sklearn.datasets import make_swiss_roll
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
X, color = make_swiss_roll(n_samples=1500, noise=0.2, random_state=42)
```

```python
# Modelo
lle = LocallyLinearEmbedding(n_neighbors=12, n_components=2, method="standard")
X_lle = lle.fit_transform(X)
```

```python
# Visualização 3D dos dados antes da aplicação de redução de dimensionalidade
fig = plt.figure(figsize=(16, 6))
ax = fig.add_subplot(1, 2, 1, projection='3d')
ax.scatter(X[:, 0], X[:, 1], X[:, 2], c=color, cmap=plt.cm.Spectral)
ax.set_title("Swiss Roll em 3D (Original)")
```

```python
# Visualização em 2D pós da aplicação de redução de dimensionalidade
fig = plt.figure(figsize=(16, 6))

ax2 = fig.add_subplot(1, 2, 2)
scatter = ax2.scatter(X_lle[:, 0], X_lle[:, 1], c=color, cmap=plt.cm.Spectral)
ax2.set_title("Swiss Roll em 2D via LLE")

plt.tight_layout()
plt.show()
```
