 # Exemplo -  01

```python
total = 10000
fraude = 200
legit = 9800

# Matriz de confusão (real x previsto)
# Modelo ruim: prevê tudo como legítimo
TP = 0   # fraude prevista como fraude
FN = 200 # fraude prevista como legítima
FP = 0   # legítima prevista como fraude
TN = 9800# legítima prevista como legítima

# Métricas principais
accuracy = (TP + TN) / total
recall_fraude = TP / (TP + FN) if (TP + FN) else 0
precision_fraude = TP / (TP + FP) if (TP + FP) else 0

confusion = {
    "TP": TP, "FN": FN, "FP": FP, "TN": TN,
    "total": total,
    "fraudes_reais": fraude,
    "legitimas_reais": legit,
    "accuracy": accuracy,
    "recall_fraude": recall_fraude,
    "precision_fraude": precision_fraude
}

confusion
```

```python
prompt = f"""
Você é um especialista em Machine Learning aplicado à detecção de fraudes.
Analise o caso abaixo e responda em português, com rigor técnico e linguagem didática.
---
Matriz de confusão:

{confusion}
---
Cenário:
- Tarefa: classificação binária (Fraude vs Legítima)
- Total de transações: {total}
- Fraudes reais: {fraude} (classe positiva rara)
- Matriz de confusão (REAL x PREVISTO):
    TP (Fraude→Fraude) = {TP}
    FN (Fraude→Legítima) = {FN}
    FP (Legítima→Fraude) = {FP}
    TN (Legítima→Legítima) = {TN}

Métricas calculadas:
- Acurácia = {accuracy:.4f}
- Recall (Fraude) = {recall_fraude:.4f}
- Precisão (Fraude) = {precision_fraude:.4f}

Tarefas:
1) Explique por que esse modelo é inadequado, mesmo com alta acurácia.
2) Indique quais métricas são mais relevantes para fraude e por quê (mencione trade-offs).
3) Sugira 5 ações práticas para melhorar o sistema (ex.: ajuste de limiar, balanceamento, métricas, validação).
4) Indique 3 riscos comuns ao interpretar métricas sem contexto (ex.: desbalanceamento, custo de erro, vazamento).
Formate a resposta com tópicos e subtítulos curtos.
"""
print(prompt)
```

```python
response = client.responses.create(
     model="gpt-4o-mini",
     input=prompt,
     temperature=0
 )

print(response.output_text)
```
