# Criação de Features Derivadas

## Exemplo 01

```python
prompt = """
Você é especialista em feature engineering.

Problema: prever preço de imóveis (regressão).
Variáveis disponíveis:
- preco (R$)
- area (m²)
- quartos
- banheiros
- bairro

Tarefa:
1) Sugira 3 features derivadas úteis.
2) Para cada feature, descreva: (a) fórmula/como calcular e (b) por que ajuda.
Responda em uma tabela.
"""
```

```python
print(prompt)
```

```python
response = client.responses.create(
    model="gpt-4o-mini",
    input=prompt,
    temperature=0
)

print(response.output_text)
```
